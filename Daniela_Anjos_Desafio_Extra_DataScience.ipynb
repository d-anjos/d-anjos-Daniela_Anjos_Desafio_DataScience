```python
# Tratamento de dados
# Identificando outlayers
# é possível filtrar os outlayers extremos, mas optei por não fazer para preservar a integridade dos dados e não criar um viés anilítico
#df_filtrado = df[(df['Profit'] > [numero]) & (df['nome_coluna'] < [numero])]

# Criando boxplots para as principais métricas destacando outlayers
# Configurar a figura
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Boxplot de Vendas
sns.boxplot(data=df, y='vendas', ax=axes[0])
axes[0].set_title('Boxplot de Vendas')
axes[0].set_ylabel('Vendas (USD)')

# Boxplot de Lucro
sns.boxplot(data=df, y='lucro', ax=axes[1])
axes[1].set_title('Boxplot de Lucro')
axes[1].set_ylabel('Lucro (USD)')

plt.tight_layout()
plt.savefig('boxplot_outliers.png', dpi=100)
plt.show()
```

```python
# Tratamento de dados
# Identificação de outliers usando IQRs

def identificar_outliers_iqr(df, coluna):
    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)
    IQR = Q3 - Q1
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    outliers = df[(df[coluna] < limite_inferior) | (df[coluna] > limite_superior)]
    return outliers, limite_inferior, limite_superior

# Aplicar nas colunas de interesse
for col in ['vendas', 'lucro']:
    outliers, li, ls = identificar_outliers_iqr(df, col)
    print(f"Outliers em {col}: {len(outliers)} ({len(outliers)/len(df)*100:.1f}% do total)")
    print(f"  Limites: [{li:.2f}, {ls:.2f}]")
    print()
```

```python
# Tratamento de dados
# Identificação de outliers usando IQRs

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, col in zip(axes, ['vendas', 'lucro']):
    sns.boxplot(data=df, y=col, ax=ax)
    _, li, ls = identificar_outliers_iqr(df, col)
    ax.axhline(li, color='red', linestyle='--', alpha=0.7, label='Limite inferior IQR')
    ax.axhline(ls, color='red', linestyle='--', alpha=0.7, label='Limite superior IQR')
    ax.set_title(f'Boxplot de {col} com limites IQR')
    ax.legend()
plt.tight_layout()
plt.savefig('outliers_com_limites.png', dpi=100)
plt.show()
```

```python
# Decisão sobre outliers: identifiquei outliers nas métricas de vendas e lucro utilizando o método do IQR (1.5x).
# Como os valores representam transações reais, optei por não removê‑los em nenhuma das análises.
# Foi adicionada a visualização limites do Intervalo Interquartil (IQR) para melhor visulização dos outlayers.
# Nenhum outlier foi excluído dos cálculos agregados.
```

```python
# Análise Exploratória de Dados (EDA)
# Gráfico de pizza vendas por Segmento de Cliente:
# O segmento Consumer não só é o maior, como também apresenta os picos sazonais mais acentuados.
# Corporate e Home Office mostram um comportamento mais estável.

segmento_group = df.groupby('segmento')['vendas'].sum().reset_index()
plt.figure(figsize=(6,6))
plt.pie(segmento_group['vendas'], labels=segmento_group['segmento'], autopct='%1.1f%%')
plt.title('Distribuição de Vendas por Segmento')
plt.tight_layout()
plt.savefig('vendas_por_segmento.png', dpi=100)
plt.show()
```

```python
# Análise Exploratória de Dados (EDA)
# Sazonalidade por Segmento (Vendas Mensais)
# Se nota que com excessão do segmento Home Office em 2014 as vendas temdem a ser mais altas em todos os seguimentos no segundo semestre.
# Avaliar o direcionamento de ações para promoção de vendas pontuais e fidelização de acordo com o comportamento de cada segmento.

# Vendas mensais por segmento
vendas_mensais_segmento = df.pivot_table(values='vendas', index=df['data_pedido'].dt.to_period('M'), columns='segmento', aggfunc='sum')
vendas_mensais_segmento.index = vendas_mensais_segmento.index.to_timestamp()

vendas_mensais_segmento.plot(figsize=(12, 6))
plt.title('Vendas Mensais por Segmento de Cliente ao Longo do Tempo')
plt.xlabel('Data do Pedido')
plt.ylabel('Vendas (USD)')
plt.grid(True)
plt.tight_layout()
plt.savefig('vendas_mensais_por_segmento.png', dpi=100)
plt.show()
```

```python
# Análise Exploratória de Dados (EDA)
# Comparando vendas e lucro ao longo do tempo
# Embora aconteça um aumento significativo de vendas no segundo semestre,
# com pico nos últimos meses do ano o lucros não acompanham de maneira proporcional.

# Agrupar vendas e lucro por mês
vendas_mensais = df.groupby(df['data_pedido'].dt.to_period('M'))['vendas'].sum()
lucro_mensal = df.groupby(df['data_pedido'].dt.to_period('M'))['lucro'].sum()

# Converter índices de período para timestamp
vendas_mensais.index = vendas_mensais.index.to_timestamp()
lucro_mensal.index = lucro_mensal.index.to_timestamp()

# Criar a figura e o eixo principal
fig, ax = plt.subplots(figsize=(12,6))

# Plotar vendas no eixo
ax.plot(vendas_mensais.index, vendas_mensais, color='blue', label='Vendas')
ax.set_xlabel('Data')
ax.set_ylabel('Valor (USD)', color='black') # Um único rótulo para o eixo Y
ax.tick_params(axis='y', labelcolor='black')

# Plotar lucro no mesmo eixo
ax.plot(lucro_mensal.index, lucro_mensal, color='green', label='Lucro')

# Título e grade
plt.title('Vendas e Lucro Mensais ao Longo do Tempo (Mesmo Eixo Y)')
plt.grid(True)
plt.legend(loc='upper left')
fig.tight_layout()

# Salvar e exibir
plt.savefig('tendencia_vendas_lucro.png', dpi=100)
plt.show()
```

```python
# Análise Exploratória de Dados (EDA)
# Perguntas: Qual categoria vende mais?
# A categoria Tecnology lidera em vendas totais.
# O valor de vendas em Furniture é maior que em Office Supplies, no entanto o lucro gerado por esta é maior sugerindo maior rentabilidade.
# Importante rever operações e avaliar se algum custo está impactando signigicativamente no lucro da categoria.

# Agrupar vendas e lucro por categoria
cat_group = df.groupby('categoria')[['vendas', 'lucro']].sum().sort_values('vendas', ascending=False)

# Gráfico de barras vendas e lucro por categoria
cat_group.plot(kind='bar', figsize=(10,5))
plt.title('Vendas e Lucro Totais por Categoria')
plt.ylabel('Valor (USD)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('vendas_por_categoria.png', dpi=100)  # Salva a imagem
plt.show()
```

```python
# Análise Exploratória de Dados (EDA)
# Análise temporal por categoria


vendas_cat_mes = df.pivot_table(values='vendas', index=df['data_pedido'].dt.to_period('M'), columns='categoria', aggfunc='sum')
vendas_cat_mes.index = vendas_cat_mes.index.to_timestamp()
vendas_cat_mes.plot(figsize=(14,6))
plt.title('Vendas Mensais por Categoria')
plt.ylabel('Vendas (USD)')
plt.grid(True)
plt.savefig('vendas_mensais_por_categoria.png')
plt.show()
```

```python
# Análise Exploratória de Dados (EDA)
# Vendas e Lucro por Subcategoria

# Agrupar vendas e lucro por subcategoria e ordenar pelo lucro (do maior para o menor)
subcat_group = df.groupby('subcategoria')[['vendas', 'lucro']].sum().sort_values('lucro', ascending=False)

# Criar o gráfico de barras
fig, ax = plt.subplots(figsize=(14, 7))
subcat_group.plot(kind='bar', ax=ax)

# Título e rótulos
ax.set_title('Vendas e Lucro Totais por Subcategoria (Ordenado por Lucro)')
ax.set_xlabel('Subcategoria')
ax.set_ylabel('Valor (USD)')
ax.tick_params(axis='x', rotation=90)

# Ajustar layout para evitar sobreposição de rótulos
plt.tight_layout()

# Salvar e exibir o gráfico
plt.savefig('vendas_lucro_por_subcategoria_por_lucro.png', dpi=100)
plt.show()
```

```python
# Análise Exploratória de Dados (EDA)
# Listando subcategorias que dão prejuízo
# As subcategorias Tables, Bookcases e Supplies estão gerando prejuízo.
# Isso merece muita atenção: rever se os custos desses produtos são mal precificados e/ou se os descontos aplicados são altos demais para essas linhas.
# Recomendo uma revisão dos preços de tabela e dos limites de desconto para essas subcategorias.

# Agrupar vendas e lucro por subcategoria
subcat_group = df.groupby('subcategoria')[['vendas', 'lucro']].sum().sort_values('vendas', ascending=False)

subcat_lucro_negativo = subcat_group[subcat_group['lucro'] < 0]

print("Subcategorias com lucro negativo:")
display(subcat_lucro_negativo)
```

```python
# As categorias Phones, Chairs, Storage e Machine possuem um discrepância grande entre vendas e lucros
# É necessário rever possibilidades de redução dos custos dessas categorias
```

```python
# Análise Exploratória de Dados (EDA)
# Existe relação entre Desconto vs. Lucro?
# Pergunta: descontos muito altos parecem causar prejuízo?
# Fica evidente uma relação negativa, à medida que o desconto aumenta, o lucro diminui e frequentemente se torna prejuízo.
# Descontos acima de 20% quase nunca geram lucro.
# A empresa deve reavaliar sua política de descontos, estabelecendo um limite máximo mais restrito.

#Gráfico de dispersão relação entre Desconto e lucro
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='desconto', y='lucro', alpha=0.5)
plt.title('Relação entre Desconto e Lucro')
plt.xlabel('Desconto')
plt.ylabel('Lucro (USD)')
plt.tight_layout()
plt.savefig('desconto_vs_lucro.png', dpi=100)
plt.show()
```

```python
# Análise Exploratória de Dados (EDA)
# Impacto do desconto no lucro nas subcategorias com prejuízo

# Pergunta: Existe relação entre Desconto vs. Lucro?
# Pergunta: descontos muito altos parecem causar prejuízo?
# Há uma relação negativa entre lucro e desconto, quanto maior o desconto menor o lucro
# Mesmo nas subcategorias lucrativas, descontos elevados podem transformar um lucro positivo em prejuízo.
# A dispersão mostra que há uma concentração de pontos no canto inferior direito (alto desconto, baixo lucro).
# Recomendo realizar análise por produto e definir limites de desconto personalizados.

# Obter a lista de subcategorias com lucro negativo
subcategorias_com_prejuizo = subcat_lucro_negativo.index.tolist()

# Filtrar o DataFrame original para essas subcategorias
df_prejuizo = df[df['subcategoria'].isin(subcategorias_com_prejuizo)]

# Agrupar por subcategoria e desconto para ver o lucro
impacto_desconto_lucro = df_prejuizo.groupby(['subcategoria', 'desconto'])['lucro'].sum().reset_index()

plt.figure(figsize=(12, 7))
sns.scatterplot(data=impacto_desconto_lucro, x='desconto', y='lucro', hue='subcategoria', style='subcategoria', s=100, palette='viridis')
plt.axhline(0, color='red', linestyle='--', linewidth=0.8) # Linha de lucro zero
plt.title('Impacto do Desconto no Lucro por Subcategoria Deficitária')
plt.xlabel('Desconto')
plt.ylabel('Lucro (USD)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Subcategoria', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('desconto_lucro_subcategorias_deficitarias.png', dpi=100)
plt.show()


```

```python
# Análise Exploratória de Dados (EDA)
# Impacto do desconto no lucro para subcategorias lucrativas

# Criar um gráfico de dispersão com as categorias que dão lucro positivo, relacionando lucro e desconto

# Identificar subcategorias com lucro positivo
subcat_lucro_positivo = subcat_group[subcat_group['lucro'] > 0]
subcategorias_com_lucro_positivo = subcat_lucro_positivo.index.tolist()

# Filtrar o DataFrame original para essas subcategorias
df_lucro_positivo = df[df['subcategoria'].isin(subcategorias_com_lucro_positivo)]

# Agrupar por subcategoria e desconto para ver o lucro
impacto_desconto_lucro_positivo = df_lucro_positivo.groupby(['subcategoria', 'desconto'])['lucro'].sum().reset_index()

plt.figure(figsize=(14, 8))
sns.scatterplot(data=impacto_desconto_lucro_positivo, x='desconto', y='lucro', hue='subcategoria', style='subcategoria', s=100, palette='Spectral')
plt.axhline(0, color='red', linestyle='--', linewidth=0.8) # Linha de lucro zero para referência
plt.title('Impacto do Desconto no Lucro para Subcategorias Lucrativas')
plt.xlabel('Desconto')
plt.ylabel('Lucro (USD)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Subcategoria', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('desconto_lucro_subcategorias_lucrativas.png', dpi=100)
plt.show()
```
